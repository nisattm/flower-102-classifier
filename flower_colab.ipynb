# Flower 102 — resumable teacher training
Upload this project folder and `flower_data` to Google Drive. Keep the `artifacts/data/folds.jsonl` file with the project. Enable a GPU runtime. Re-running the last cell resumes from the previous completed epoch in Drive.

```python
from google.colab import drive
drive.mount('/content/drive')
from pathlib import Path
project = Path('/content/drive/MyDrive/Flower_Project')
data = Path('/content/drive/MyDrive/flower_data')
assert (project / 'train.py').exists(), project
assert (project / 'artifacts/data/folds.jsonl').exists()
assert (data / 'train').exists(), data
import shutil
local_data = Path('/content/flower_data')
if not (local_data / 'train').exists():
    shutil.copytree(data, local_data, dirs_exist_ok=True)
data = local_data
```

```python
import subprocess, torch
print('PyTorch', torch.__version__, 'CUDA', torch.version.cuda, 'GPU', torch.cuda.get_device_name(0) if torch.cuda.is_available() else None)
assert torch.cuda.is_available(), 'Select a GPU runtime first'
subprocess.run(['python', '-m', 'pip', 'install', '-q', 'timm==1.0.25'], check=True)
```

```python
import os, subprocess
env = os.environ.copy()
env['HF_HOME'] = str(project / 'artifacts/hf_cache')
cmd = ['python', 'train.py', '--data', str(data), '--folds', str(project / 'artifacts/data/folds.jsonl'), '--manifest', str(project / 'artifacts/data/manifest.jsonl'), '--run', str(project / 'artifacts/runs/teacher_f0'), '--model', 'teacher', '--fold', '0', '--batch', '2', '--workers', '2']
print(' '.join(cmd))
subprocess.run(cmd, cwd=project, env=env, check=True)
```
